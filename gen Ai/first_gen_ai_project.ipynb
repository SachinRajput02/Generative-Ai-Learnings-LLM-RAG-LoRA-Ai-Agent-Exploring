```python
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser
from secret_key import geminiapi_key

# prompt
prompt_template_name = Pro mptTemplate(
    input_variables=["occupation"],
    template="give me the list of indian {occupation}"
)

# LLM
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", google_api_key=geminiapi_key)

# Output parser
parser = StrOutputParser()

# LCEL chain
chain = prompt_template_name | llm | parser

# run
result = chain.invoke({"occupation": "cricketers"})
print(result)
```
