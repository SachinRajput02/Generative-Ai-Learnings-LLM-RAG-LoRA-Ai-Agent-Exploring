To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
import os
from dotenv import load_dotenv
from openai import OpenAI
import ollama
from IPython.display import display, Markdown
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2:1b'
```

```python
# set up environment
load_dotenv(override=True)

# OpenAI client
openai = OpenAI()

# Verify that Ollama is running and the model is available
print("Environment loaded.")
```

```python
question = """
Explain the following Python code step by step. Describe how async/await
works, why asyncio.gather() is used, and what the output represents.

import asyncio

async def fetch_data(item):
    print(f"Fetching {item}")
    await asyncio.sleep(1)
    return f"Data for {item}"

async def main():
    results = await asyncio.gather(
        fetch_data("users"),
        fetch_data("products"),
        fetch_data("orders")
    )
    print(results)

asyncio.run(main())
"""
```

```python
# Get gpt-4o-mini to answer, with streaming

stream = openai.chat.completions.create(
    model=MODEL_GPT,
    messages=[
        {
            "role": "system",
            "content": (
                "You are a helpful technical tutor. "
                "Explain code clearly, step by step, with examples."
            )
        },
        {
            "role": "user",
            "content": question
        }
    ],
    stream=True
)

print("GPT-4o-mini answer:\n")

for chunk in stream:
    content = chunk.choices[0].delta.content
    if content:
        print(content, end="", flush=True)

print()
```

```python
# Get Llama 3.2 to answer

response = ollama.chat(
    model=MODEL_LLAMA,
    messages=[
        {
            "role": "system",
            "content": (
                "You are a helpful technical tutor. "
                "Explain code clearly, step by step, with examples."
            )
        },
        {
            "role": "user",
            "content": question
        }
    ]
)

print("Llama 3.2 answer:\n")
print(response["message"]["content"])
```
